## 1) Imports — libraries you will use

- **pandas**: load tables (rows = examples, columns = features + label).
- **sklearn**: split data, vectorize text, train Random Forest, tune, evaluate.
- **matplotlib / seaborn**: plots for EDA and the confusion matrix.

````python
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    accuracy_score,
    classification_report,
    f1_score,
)
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

# --- EDIT THESE TO MATCH YOUR KAGGLE FILE AND COLUMNS ---
ROOT = Path.cwd()
CSV_PATH = ROOT / "ml" / "data" / "YOUR_FILE.csv"  # change filename

# If your label is text (e.g. "Italian"), sklearn will encode it automatically when you pass y to .fit().
TEXT_COLUMN = "ingredients"  # change: one text column used as input (or see notes below)
LABEL_COLUMN = "cuisine"  # change: the column you want to predict (supervised target)

RANDOM_STATE = 42
TEST_SIZE = 0.2  # 20% held out for final evaluation only
````

## 2) Load the data — each row is one training example

`read_csv` loads your Kaggle export. If the file is tab-separated, use `sep="\t"`.

````python
if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Missing {CSV_PATH}. Download your Kaggle dataset into ml/data/ and update CSV_PATH."
    )

df = pd.read_csv(CSV_PATH)
print("Shape (rows, columns):", df.shape)
df.head()
````

## 3) Exploratory Data Analysis (EDA) — understand before you model

- **Missing values**: drop or fill.
- **Class balance**: if one cuisine is 95% of rows, accuracy can be misleading; use F1 / confusion matrix.
- **Duplicates**: remove if needed.

````python
df.info()
df[LABEL_COLUMN].value_counts()
````

````python
plt.figure(figsize=(10, 4))
order = df[LABEL_COLUMN].value_counts().index
sns.countplot(data=df, y=LABEL_COLUMN, order=order, color="steelblue")
plt.title("How many examples per class?")
plt.tight_layout()
plt.show()
````

## 4) Define X and y — supervised learning needs inputs + correct answers

- **X**: features the model is *allowed* to see at prediction time.
- **y**: the label you want to predict (the "answer key" for training).

This template uses **one text column** → **TF-IDF** (turns text into numeric features). If you have **only numeric** columns, replace the pipeline later with a scaler + forest on those columns only (ask your TA or adapt from sklearn docs).

````python
if TEXT_COLUMN not in df.columns or LABEL_COLUMN not in df.columns:
    raise KeyError(
        f"Check column names. Available: {list(df.columns)}. "
        f"Looking for TEXT_COLUMN={TEXT_COLUMN!r}, LABEL_COLUMN={LABEL_COLUMN!r}"
    )

use = df[[TEXT_COLUMN, LABEL_COLUMN]].dropna()
use = use[use[TEXT_COLUMN].astype(str).str.len() > 2]

X = use[TEXT_COLUMN]
y = use[LABEL_COLUMN]

print("Examples after dropping missing/short text:", len(use))
````

## 5) Train / test split — never tune on the test set

- **Training** set: fit the model.
- **Test** set: used **once** at the end for honest metrics (what you report on slides).

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y
)
print("Train:", len(X_train), " Test:", len(X_test))
````

## 6) Pipeline — preprocessing + model in one object

**Why:** so cross-validation and test evaluation apply the *same* steps (e.g. TF-IDF fit on train only, not on test — avoids leakage).

````python
pipeline = Pipeline(
    steps=[
        ("tfidf", TfidfVectorizer(max_features=5000, ngram_range=(1, 2))),
        (
            "clf",
            RandomForestClassifier(
                random_state=RANDOM_STATE,
                class_weight="balanced",
                n_jobs=-1,
            ),
        ),
    ]
)

pipeline
````

## 7) Fine-tune with GridSearchCV — try hyperparameters on *train* only

This uses 3-fold **cross-validation** inside the training split. Smaller grids run faster; expand if you have time.

````python
param_grid = {
    "clf__n_estimators": [100, 200],
    "clf__max_depth": [None, 20, 40],
    "clf__min_samples_leaf": [1, 2],
}

search = GridSearchCV(
    pipeline,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=3,
    n_jobs=-1,
    verbose=1,
)

search.fit(X_train, y_train)
print("Best params:", search.best_params_)
print("Best CV f1_macro:", search.best_score_)
````

## 8) Final evaluation on the held-out test set — numbers for your slides

Use **`search.best_estimator_`** (already trained on all of `X_train`). Report metrics on **`X_test, y_test`** only.

````python
best = search.best_estimator_
y_pred = best.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Macro F1:", f1_score(y_test, y_pred, average="macro"))
print()
print(classification_report(y_test, y_pred))
````

````python
fig, ax = plt.subplots(figsize=(8, 8))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, ax=ax, xticks_rotation=90)
plt.title("Confusion matrix (test set)")
plt.tight_layout()
plt.show()
````

### ROC curve

**Binary classification:** `RocCurveDisplay.from_predictions` works directly.

**Many classes (multiclass):** either show **one-vs-rest** ROC per class (busy), or skip ROC and rely on confusion matrix + macro F1 — ask your instructor what they prefer.

````python
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import RocCurveDisplay
from sklearn.multiclass import OneVsRestClassifier

classes = sorted(pd.unique(y))
if len(classes) == 2:
    le = LabelEncoder()
    y_test_bin = le.fit_transform(y_test)
    # RandomForest.predict_proba for binary gives shape (n, 2) — use probability of positive class
    pos_idx = list(le.classes_).index(classes[1]) if len(le.classes_) == 2 else 1
    proba = best.predict_proba(X_test)[:, pos_idx]
    RocCurveDisplay.from_predictions(y_test_bin, proba)
    plt.title("ROC (test set, binary)")
    plt.show()
else:
    print(f"Skipping default ROC plot: {len(classes)} classes. Use multiclass ROC or report macro-F1 + confusion matrix.")
````

## 9) Save the trained pipeline — "deploy" for later or for demo

Saving the **whole pipeline** (TF-IDF + forest) ensures inference uses the same preprocessing.

````python
out_path = ROOT / "ml" / "data" / "recipe_rf_pipeline.joblib"
joblib.dump(best, out_path)
print("Saved to:", out_path)
````
